# Opis problemu
Niespodziewane odejście wykwalifikowanego pracownika z pracy bywa problematyczne dla pracodawcy, który ma niewiele czasu na znalezienie nowego pracownika na jego miejsce. W pliku **HR_Analytics.csv** znajdują się informacje o 1470 pracownikach, wraz z informacją czy zdecydowali się oni opuścić swojego pracodawce. Zadania poniżej mają na celu ustalenie, jakie czynniki mogłą wpływać na podejmowanie przez pracownika decyzji o zmianie pracodawcy.

Dodatkowe informacje o zbiorze danych: https://www.kaggle.com/datasets/thedevastator/employee-attrition-and-factors

## Zadania
1. Stwórz wykres (liniowy) przedstawiający współczynnik odejść w zależności od wieku z podziałem według płci. (0.5 pkt)
2. Stwórz wykres (parallel_categories), zawierający informację o osobach zarabiających przynajmniej 100k rocznie, przedstawiający stosunek osób, które zdecydowały się odejść, do osób, które pozostały. Względnij zmienne zdefiniowanych przez JobInvolvement, JobLevel, Education, JobSatisfaction, MaritalStatus, StockOptionLevel, WorkLifeBalance. (0.5 pkt)
3. Stwórz wykres (punktowy) przedstawiający zależność pomiędzy zarobkami (y - MonthlyIncome), a latami przeprowacowanymi u danego pracodawcy (x - YearsAtCompany) z podziałem na staniowiska na których pracują (kolor, uwzględnij tylko 5 najliczniejszych z JobRole) i dotychczasowym doświadczeniem ( wielkość punktu - TotalWorkingYears). Dodaj do wykresu linie trendu dla każdego stanowiska. ( 1 pkt)w
