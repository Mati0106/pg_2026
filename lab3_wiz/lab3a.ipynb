# Plotly wprowadzenie
Biblioteka plotly jest interaktywną, otwartą biblioteką do tworzenia wykresów, która obsługuje ponad 40 unikalnych typów wykresów obejmujących szeroki zakres zastosowań statystycznych, finansowych, geograficznych, naukowych i trójwymiarowych.

Zbudowana na bibliotece Plotly JavaScript (plotly.js), plotly umożliwia użytkownikom Pythona tworzenie interaktywnych wizualizacji internetowych, które mogą być wyświetlane w notatnikach Jupytera, zapisywane w samodzielnych plikach HTML lub statycznych plikach graficznych (jpeg,png,svg,pdf,..), albo serwowane jako część aplikacji internetowych zbudowanych w czystym Pythonie przy użyciu Dash'a.

Korzystanie z bilbioteki Plotly należy zaczać od zainstalowania potrzebnych pakietów i zaimportowania modułu. 

## Figures As Dictionaries
Podobnie do matplotlib, plotly również posiada obiekt figure, który będzie zawierał wszystkie informację o wykresie. Figure w plotly posiada strukturę drzewistą, z trzema gałęziami na pierwszym poziomie: data, layout i frames. 

Strukturę potrzebną do stworzenia wykresu można zapisać bezpośrednio korzystając z typów wbudowanych, tak jak na poniższym przykładzie:

```python
fig = dict({
    "data": [{"type": "bar",
              "x": [1, 2, 3],
              "y": [1, 3, 2]}],
    "layout": {"title": {"text": "A Figure Specified By Python Dictionary"}}
})

import plotly.io as pio
pio.show(fig)
```


## Figures as Graph Objects



```python
import plotly.graph_objects as go

fig = go.Figure(
    data=[go.Bar(x=[1, 2, 3], y=[1, 3, 2])],
    layout=go.Layout(
        title=go.layout.Title(text="A Figure Specified By A Graph Object")
    )
)

fig.show()
```

Listę wszystkich obiektów w module graph_object można znaleźć w dokumentacji (https://plotly.com/python-api-reference/plotly.graph_objects.html#graph-objects).

### Graph object from dictionary

```python
import plotly.graph_objects as go

dict_of_fig = dict({
    "data": [{"type": "bar",
              "x": [1, 2, 3],
              "y": [1, 3, 2]}],
    "layout": {"title": {"text": "A Figure Specified By A Graph Object With A Dictionary"}}
})

fig = go.Figure(dict_of_fig)

fig.show()
```

## Converting between types

```python
import plotly.graph_objects as go

fig = go.Figure(
    data=[go.Bar(x=[1, 2, 3], y=[1, 3, 2])],
    layout=go.Layout(height=600, width=800)
)

fig.layout.template = None # to slim down the output

print("Dictionary Representation of A Graph Object:\n\n" + str(fig.to_dict()))
print("\n\n")
print("JSON Representation of A Graph Object:\n\n" + str(fig.to_json()))
print("\n\n")
```

## Difference between graph_object and plotly.express

```python
import pandas as pd

df = pd.DataFrame({
  "Fruit": ["Apples", "Oranges", "Bananas", "Apples", "Oranges", "Bananas"],
  "Contestant": ["Alex", "Alex", "Alex", "Jordan", "Jordan", "Jordan"],
  "Number Eaten": [2, 1, 3, 1, 3, 2],
})


# Plotly Express

import plotly.express as px

fig = px.bar(df, x="Fruit", y="Number Eaten", color="Contestant", barmode="group")
fig.show()


# Graph Objects

import plotly.graph_objects as go

fig = go.Figure()
for contestant, group in df.groupby("Contestant"):
    fig.add_trace(go.Bar(x=group["Fruit"], y=group["Number Eaten"], name=contestant,
      hovertemplate="Contestant=%s<br>Fruit=%%{x}<br>Number Eaten=%%{y}<extra></extra>"% contestant))
fig.update_layout(legend_title_text = "Contestant")
fig.update_xaxes(title_text="Fruit")
fig.update_yaxes(title_text="Number Eaten")
fig.show()
```

## Adding traces

```python
fig = go.Figure() 

fig.add_scatter(y=[4, 2, 1], mode="lines")
fig.add_bar(y=[2, 1, 3])

fig.show()
```

Trace'y można również dodawać do wcześniej utworzonych wykresów.


```python
import plotly.express as px

df = px.data.iris()

fig = px.scatter(df, x="sepal_width", y="sepal_length", color="species",
                 title="Using The add_trace() method With A Plotly Express Figure")

fig.add_trace(
    go.Scatter(
        x=[2, 4],
        y=[4, 8],
        mode="lines",
        line=go.scatter.Line(color="gray"),
        showlegend=False)
)
fig.show()
```
Alternatywnie dla **add_trace** kazdy typ trace'u posiada swój własny alias **add_\<trace_name\>**.

```python
import plotly.express as px

df = px.data.stocks(indexed=True)
fig = px.line(df)
fig.add_hline(y=1, line_dash="dot",
              annotation_text="Jan 1, 2018 baseline", 
              annotation_position="bottom right",
              annotation_font_size=20,
              annotation_font_color="blue"
             )
fig.add_vrect(x0="2018-09-24", x1="2018-12-18", 
              annotation_text="decline", annotation_position="top left",
              annotation=dict(font_size=20, font_family="Times New Roman"),
              fillcolor="green", opacity=0.25, line_width=0)
fig.show()
```

## Make subplot

```python
from plotly.subplots import make_subplots

fig = make_subplots(rows=1, cols=2)

fig.add_trace(go.Scatter(y=[4, 2, 1], mode="lines"), row=1, col=1)
fig.add_trace(go.Bar(y=[2, 1, 3]), row=1, col=2)

fig.show()
```

## Updating data after figure creation
```python
import plotly.graph_objects as go

fig = go.Figure(data=go.Bar(x=[1, 2, 3], y=[1, 3, 2]))

fig.update_layout(title_text="Using update_layout() With Graph Object Figures",
                  title_font_size=30)

fig.show()
```

```python

fig.update_layout({"title": {"text": "update_layout() Syntax Example",
                             "font": {"size": 30}}}) # dictionary

fig.update_layout(dict(title=dict(text="update_layout() Syntax Example"),
                             font=dict(size=30)))

fig.update_layout(title=go.layout.Title(text="update_layout() Syntax Example",
                                        font=go.layout.title.Font(size=30)))

fig.update_layout(title_text="update_layout() Syntax Example",
                  title_font_size=30) # magic underscore

fig.update_layout(title_text="update_layout() Syntax Example",
                  title_font=dict(size=30))

fig.layout.title.text = "update_layout() Syntax Example"
fig.layout.title.font.size = 30

```
