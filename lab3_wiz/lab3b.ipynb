# Plotly express

Plotly składa się z kilku modułów pozwalających budować i modyfikować różne części wykresu. Na tych zajęciach skupimy się na module **express**, oferjącym wysokopoziomowy dostęp do wielu podstawowych typów wykresów i konfiguracji. 

```python
import plotly.express as px
```

Wykresy dostępne w module **express** możmy podzielić na kilka kategorii:
 - *podstawowe*: scatter, line, area, bar, funnel, timeline
 - *wycinkowe*: pie, sunburst, treemap, icicle, funnel_area
 - *rozkłady jednowymiarowe*: histogram, box, violin, strip, ecdf
 - *rozkłady dwuwymiarowe*: density_heatmap, density_contour
 - *macierze lub zdjęcia (wykresy rastrowe)*: imshow
 - *trójwymiarowe*: scatter_3d, line_3d
 - *wielowymiarowe*: scatter_matrix, parallel_coordinates, parallel_categories
 - *mapy kafelkowe/koncentracji*: scatter_mapbox, line_mapbox, choropleth_mapbox, density_mapbox
 - *mapy punktowe/liniowe*: scatter_geo, line_geo, choropleth
 - *polarne*: scatter_polar, line_polar, bar_polar
 - *trójskładnikowe*: scatter_ternary, line_ternary

Pełna dokumentacja modułu express: https://plotly.com/python-api-reference/plotly.express.html

Moduł express tworzy za nas obiekt *Figure*, który jest podstawowym obiektem reprezentującym wszystkie wykresy w Plotly i uzupełnia za nas wszystkie wymagane dla tego obiektu parametry.   
 ```python
fig = px.line(x=["a","b","c"], y=[1,3,2])
print(fig)
```

Modyfikowanie takiego wykresu może odbywać sie poprzez bezpośrednie modyfikowanie jego atrybutów, ale wygodniejszą i bezpieczniejszą opcją się wykorzystanie przygotowanych w tym celu metod i argumentów funkcji tworzącej wykres.

 - zmiana podpisu osi
```python
fig = px.scatter(x=t, y=np.sin(t), labels={'x':'t', 'y':'sin(t)'})
```
 - zmiana/dodanie tytułu wykresu
```python
fig = px.scatter(x=t, y=np.sin(t), title='New Chart Title')
```
 - zmiana szablonu
 ```python
fig = px.scatter(x=t, y=np.sin(t), template="simple_white")
```
 - zmiana szerokości i wysokości
 ```python
fig = px.scatter(x=t, y=np.sin(t), width=600, height=400)
```
Wszystkie opcje dla konkretnego typu wykresu można znaleźć w dokumentacji Plotly (https://plotly.com/python-api-reference/plotly.express.html). Dodatkowo, (niemal) każdy wykres posiada możliwość dodania do niego dodatkowych informacji w postaci kolorów, wielkości punktów, tekstu czy w dodatkowym oknie tekstowym. Wszystkie te informacje dodamy używając odpowiednich argumentów w funkcji tworzącej:

```python
import plotly.express as px
df = px.data.iris()
fig = px.scatter(df, x="sepal_width", y="sepal_length", size = "petal_length", color="species", marginal_y="violin", marginal_x="box", trendline="ols", template="simple_white")
fig.show()
```

Modifikowanie wykresu już po utworzeniu obiektu *Figure* jest również możliwe i odbywa się przy pomocy metod tego obiektu. Pełna dokumentacja tych metod znajduje się na stronie https://plotly.com/python-api-reference/generated/plotly.graph_objects.Figure.html#plotly.graph_objects.Figure.update_layout .

Wykresy stworzone przy pomocy Plotly można zapisać w wersji statycznej (png, jpeg, svg, pdf, ...) jak i dynamicznej (webowej). 
Zapis wersji statycznej wykonywany jest przez Kaleido ( lub Orca - deprecated ). Kaleido jest wieloplatformową biblioteką do generowania statycznych obrazów (png, svg, pdf,...) dla internetowych bibliotek wizualizacji. Projekt ten został utworzony na potrzeby Plotly i jego główną zaletą jest minimalna liczba dodatkowych zależności.
Po zainstalowaniu Kaleido możemy zaczać używać metody *write_image* na dowolnym obiekcie **plotly.Figure**, aby zapisać wykres na naszym komputerze. Format zostanie automatycznie rozpoznany na podstawie rozszerzenia jakie podamy przy nazwie pliku. 

```python
fig.write_image("images/fig1.png")
```

Wersje dynamiczną mozna zapisać bez instalowania dodatkowych pakietów uzywając tej samej metody i rozszerzenia **.html**. Pliki z rozszerzeniem HTML możemy następnie otworzyć korzystając z dowolnej przeglądarki internetowej.

```python
fig.write_html("images/fig1.html")
```
